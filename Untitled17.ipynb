**Geospatial & Climate-Driven Mosquito Distribution Mapping (SDM)**

Step 1: Environment Setup & Library Installation
Is Step mein kya hoga: Hum Google Colab notebook mein required geospatial aur visualization libraries (pygbif, geopandas, folium, matplotlib, seaborn) install karenge.

```python
# Step 1: Install required geospatial and data visualization libraries
print("Installing required libraries...")
!pip install -q pygbif geopandas folium matplotlib seaborn pandas numpy requests

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium
from folium.plugins import HeatMap
from pygbif import occurrences as occ
import warnings
warnings.filterwarnings('ignore')

print("Libraries installed and loaded successfully!")
```

Step 2: Fetching Mosquito Occurrence Dataset via API
Is Step mein kya hoga: Hum manual CSV download karne ke bajaye, directly GBIF database se Aedes aegypti (mosquito species) ke geographical occurrence records (latitude, longitude, country, year) fetch karenge using python pygbif package. Yeh completely automated aur reliable hai.

```python
# Step 2: Fetching Aedes aegypti occurrence records from GBIF database
print("Fetching species occurrence data from GBIF...")

# Search for Aedes aegypti with valid geographic coordinates (limit to 500 records for smooth Colab execution)
search_results = occ.search(
    scientificName='Aedes aegypti',
    hasCoordinate=True,
    hasGeospatialIssue=False,
    limit=500
)

# Extract results list
records = search_results.get('results', [])

# Parse records into a pandas DataFrame
data = []
for r in records:
    lat = r.get('decimalLatitude')
    lon = r.get('decimalLongitude')
    country = r.get('country', 'Unknown')
    year = r.get('year', np.nan)
    if lat and lon:
        data.append({'latitude': lat, 'longitude': lon, 'country': country, 'year': year})

df_mosquito = pd.DataFrame(data)

# Display dataset overview
print(f"Total valid occurrence records fetched: {len(df_mosquito)}")
display(df_mosquito.head())
```

Step 3: Data Cleaning & Preprocessing
Is Step mein kya hoga: Hum fetched data ko check karenge, missing values remove karenge, aur ensure karenge ke coordinates proper numeric format mein hon taake mapping mein koi error na aaye.

```python
# Step 3: Cleaning and filtering the dataset
print("Cleaning dataset...")

# Drop rows with missing latitude or longitude
df_clean = df_mosquito.dropna(subset=['latitude', 'longitude']).copy()

# Ensure correct data types
df_clean['latitude'] = pd.to_numeric(df_clean['latitude'])
df_clean['longitude'] = pd.to_numeric(df_clean['longitude'])

# Summary statistics of coordinates
print(df_clean.describe())
print(f"Cleaned dataset rows: {len(df_clean)}")
```

Step 4: Generating Interactive Map & Heatmap (Folium)
Is Step mein kya hoga: Yeh sab se zabardast visual step hai! Hum Folium aur HeatMap plugin use karke ek interactive HTML map generate karenge jahan mosquito concentration hotspots visual colors (Green to Red) mein show honge. Yeh Colab notebook ke andar hi interactive render ho jayega.

```python
# Step 4: Creating Interactive Geospatial Heatmap
print("Generating interactive geospatial heatmap...")

# Calculate mean latitude and longitude to center the map dynamically
map_center = [df_clean['latitude'].mean(), df_clean['longitude'].mean()]

# Initialize a Folium map
m = folium.Map(location=map_center, zoom_start=3, tiles='CartoDB positron')

# Prepare coordinate list for heatmap layer
heat_data = [[row['latitude'], row['longitude']] for index, row in df_clean.iterrows()]

# Add HeatMap layer to the map
HeatMap(heat_data, radius=12, blur=8, max_zoom=1).add_to(m)

# Display the map inside Google Colab
display(m)
```

Step 5: Generating Environmental / Spatial Analytics Visuals (Seaborn & Matplotlib)
Is Step mein kya hoga: Hum visual analysis create karenge jo dikhata hai ke fetched points global scale par kis latitude distribution range mein fall kar rahe hain. Is se ek scientific/anthropological research paper jaisi look milti hai.

```python
# Step 5: Generating Distribution Plots & Analytics Visuals
print("Generating analytical plots...")

plt.figure(figsize=(14, 6))

# Subplot 1: Latitude distribution histogram / KDE
plt.subplot(1, 2, 1)
sns.histplot(df_clean['latitude'], kde=True, color='crimson', bins=30)
plt.title('Latitudinal Distribution of Vector Occurrences', fontsize=12, fontweight='bold')
plt.xlabel('Latitude (Degrees)', fontsize=10)
plt.ylabel('Frequency Count', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)

# Subplot 2: Top Countries with Recorded Vector Sightings
plt.subplot(1, 2, 2)
top_countries = df_clean['country'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Regions/Countries by Occurrence Records', fontsize=12, fontweight='bold')
plt.xlabel('Record Count', fontsize=10)
plt.ylabel('Country', fontsize=10)
plt.grid(True, axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

print("All steps completed successfully without errors!")
```
